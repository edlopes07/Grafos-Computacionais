# **Lista de Adjacência**

```python
grafo = {}
while True:
  x = input("""
LISTA DE ADJACÊNCIA\n
[1] Inserir vértice
[2] Remover vértice
[3] Inserir aresta
[4] Remover aresta
[5] Calcular grau de uma aresta
[6] Verificar possível aresta (de x para y)
[7] Listar vizinhos de um vértice
[8] Verificar se um determinado percurso é possível (de x para y)
[9] Exibir grafo
[0] Sair\n
--> """)

  match(int(x)):
    case 1:
      vertice = input(f"\nInsira um vértice: ")
      grafo[vertice] = []
      print(grafo)

    case 2:
      vertice = input(f"\nInsira o vértice a ser removido: ")
      if vertice in grafo:
        for v in grafo:
          if vertice in grafo[v]:
            grafo[v].remove(vertice)
        del grafo[vertice]
      else:
        print("O vértice não existe.")
      print(grafo)

    case 3:
      vertice1 = input(f"\nEscolha o vértice nº1: ")
      vertice2 = input(f"\nEscolha o vértice nº2: ")
      if vertice1 not in grafo or vertice2 not in grafo:
        print("Um ou mais vértices não existem.")
      else:
          grafo[vertice1].append(vertice2)
          grafo[vertice2].append(vertice1)
      print(grafo)

    case 4:
      vertice1 = input(f"\nEscolha o vértice nº1: ")
      vertice2 = input(f"\nEscolha o vértice nº2: ")
      if vertice1 not in grafo or vertice2 not in grafo:
        print("Um ou mais vértices não existem.")
      else:
          grafo[vertice1].remove(vertice2)
          grafo[vertice2].remove(vertice1)
      print(grafo)

    case 5:
      vertice = input(f"\nEscolha o vértice: ")
      if vertice in grafo:
        print(f"O grau do vértice {vertice} é {len(grafo[vertice])}")
      else:
        print("O vértice não existe.")

    case 6:
      vertice1 = input(f"\nEscolha o vértice nº1: ")
      vertice2 = input(f"\nEscolha o vértice nº2: ")
      if vertice1 not in grafo or vertice2 not in grafo:
        print("Um ou mais vértices selecionados não existem.")
      elif vertice2 in grafo[vertice1]:
        print("A aresta existe.")
      else:
        print("A aresta não existe.")

    case 7:
      vertice = input(f"\nEscolha o vértice: ")
      if vertice in grafo:
        print(f"Os vizinhos do vértice {vertice} são: {grafo[vertice]}")
      else:
        print("O vértice não existe.")

    case 8:
      vertice1 = input(f"\nEscolha o vértice inicial: ")
      vertice2 = input(f"\nEscolha o vértice final: ")
      if vertice1 not in grafo or vertice2 not in grafo:
        print("Um ou mais vértices não existem.")
      else:
        visitados = [vertice1]
        fila = [vertice1]
        possivel = False

        while len(fila) > 0:
          atual = fila.pop(0)

          if atual == vertice2:
            possivel = True
            break

          for vizinho in grafo[atual]:
            if vizinho not in visitados:
              visitados.append(vizinho)
              fila.append(vizinho)

        if possivel:
          print(f"Existe um percurso entre '{vertice1}' e '{vertice2}'.")
        else:
          print(f"Não existe percurso entre '{vertice1}' e '{vertice2}'.")
    case 9:
      print(grafo)

    case 0:
      break
```

# **Matriz de Adjacência**

```python
matriz = []
vertices = []

def exibirMatriz():
  print(vertices)
  for linha in matriz:
    print(linha)

def manipularVertice(v, tipo):
    if tipo == 1:
        if v in vertices:
            print(f"{v} já existe!")
            return

        vertices.append(v)

        for linha in matriz:
            linha.append(0)

        matriz.append([0] * len(vertices))
        return matriz

    if tipo == 2:
      if v not in vertices:
        print(f"{v} não existe!")
        return

      indice = vertices.index(v)
      vertices.remove(v)
      matriz.pop(indice)

      for linha in matriz:
        del linha[indice]
      return matriz

def manipularAresta(v1, v2, tipo):
  if v1 not in vertices or v2 not in vertices:
    print(f"Não é possivel relacionar [{v1}][{v2}]!\nVerifique se os vértices existem.")
    return
  if tipo == 1:
    indice1 = vertices.index(v1)
    indice2 = vertices.index(v2)

    matriz[indice1][indice2] = 1
    matriz[indice2][indice1] = 1
    return matriz

  if tipo == 2:
    indice1 = vertices.index(v1)
    indice2 = vertices.index(v2)

    matriz[indice1][indice2] = 0
    matriz[indice2][indice1] = 0
    return matriz

def calcularGrau():
  if not matriz:
    return "A Matriz Adjacência está vazia."

  calculados = {}
  for v in vertices:
    indice = vertices.index(v)
    grau = 0
    for i in matriz[indice]:
      if i == 1:
        grau += 1
    calculados[v] = [grau]
  return calculados

def listarVizinhos(v):
  if v not in vertices:
    return f"{v} não existe!"

  coluna = -1
  vizinhos = []
  indice = vertices.index(v)

  for i in matriz[indice]:
    coluna += 1
    if i == 1:
      vizinhos.append(vertices[coluna])
  print(f"Vizinhos de {v}:")
  return vizinhos


while True:
  x = input("""
MATRIZ DE ADJACÊNCIA
[1] Manipular vértices
[2] Manipular arestas
[3] Exibir grau de cada vértice
[4] Verificar possível aresta (de x para y)
[5] Listar vizinhos de um vértice
[6] Verificar se um determinado percurso é possível (de x para y)
[0] Sair\n
--> """)
  match(int(x)):
    case 1:
      y = input("""
  MANIPULAÇÃO DE VÉRTICES
  [1] Adicionar vértice
  [2] Remover vértice
  [0] Voltar\n
  --> """)
      match(int(y)):
        case 1:
          try:
            vertice = int(input("\nInsira um vértice: "))
          except:
            print("O vértice deve ser um número inteiro.")
            continue
          manipularVertice(vertice, 1)
          exibirMatriz()

        case 2:
          try:
            print(f"\nEscolha um vértice para remover {vertices}")
            vertice = int(input("--> "))
          except:
            print("O vértice deve ser um número inteiro.")
            continue
          manipularVertice(vertice, 2)
          exibirMatriz()

        case 0:
          continue

    case 2:
      y = input("""
MANIPULAÇÃO DE ARESTAS
[1] Adicionar aresta
[2] Remover aresta
[0] Voltar\n
--> """)
      match(int(y)):
        case 1:
          try:
            print(f"\nEscolha os vértices que irão compor a aresta ({vertices})")
            vertice1 = int(input("--> "))
            vertice2 = int(input("--> "))
          except:
            print("Os vértices devem ser um número inteiro.")
            continue
          manipularAresta(vertice1, vertice2, 1)
          exibirMatriz()

        case 2:
          try:
            print(f"\nEscolha os vértices que compõem a aresta ({vertices})")
            vertice1 = int(input("--> "))
            vertice2 = int(input("--> "))
          except:
            print("Os vértices devem ser um número inteiro.")
            continue
          manipularAresta(vertice1, vertice2, 2)
          exibirMatriz()

        case 0:
          continue

    case 3:
      print(calcularGrau())

    #case 4:

    case 5:
      try:
            print(f"\nEscolha um vértice {vertices}")
            vertice = int(input("--> "))
      except:
        print("O vértice deve ser um número inteiro.")
        continue
      print(listarVizinhos(vertice))

    #case 6:

    case 0:
      break
```
